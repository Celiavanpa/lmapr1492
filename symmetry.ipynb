```python
import pymatgen as mg
from pymatgen.ext.matproj import MPRester
from pymatgen.symmetry.analyzer import SpacegroupAnalyzer
from jupyter_jsmol import JsmolView
from IPython.display import display

API_KEY = "zA1YwwFoS7C8RQ3ZpVn2lK2ClRoSNTFM"
mp_id = "mp-27869"  

with MPRester(API_KEY) as m:
    structure = m.get_structure_by_material_id(mp_id)

# Analyse symétrie
symmetry_analyzer = SpacegroupAnalyzer(structure)
space_group = symmetry_analyzer.get_space_group_symbol()
equivalent_positions = symmetry_analyzer.get_symmetry_operations()

print(f"Groupe d'espace : {space_group}")
print("Opérations de symétrie (3 premières) :")
for op in equivalent_positions[:3]:  
    print(op)

# Sélection de trois atomes différents 
atoms = structure.sites[:3]  
transformed_structures = []

# Application opérations de symétrie sur les atomes
for i, op in enumerate(equivalent_positions[:3]):
    transformed_structure = structure.copy()
    transformed_structure.apply_operation(op)
    transformed_structures.append(transformed_structure)
    print(f"Transformation {i+1} appliquée")

# Visualisation des structures transformées avec Jsmol
jsmol_views = []
for transformed_structure in transformed_structures:
    jsmol_view = JsmolView.from_str(transformed_structure.to(fmt="cif"), "cif")
    jsmol_views.append(jsmol_view)

# Affichage des vues 3D des structures transformées
for jmol in jsmol_views:
    display(jmol)
```
