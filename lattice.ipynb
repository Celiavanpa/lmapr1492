```python
from mp_api.client import MPRester

# Initialisation de MPRester avec ta clé d'API
mpr = MPRester("ER5V8i0JUw7k31Nk5IpkvogP61BDz9cw")

# Récupérer la structure à partir du MP-id
structure = mpr.get_structure_by_material_id("mp-27869")

# Afficher la structure
print(structure)
```

```python
#vecteurs de la maille directe
lattice = structure.lattice

a1, a2, a3 = lattice.matrix
print("Vecteurs de la maille directe:", a1, a2, a3)

#vecteurs de la maille réciproque
reciprocal_lattice = lattice.reciprocal_lattice

b1, b2, b3 = reciprocal_lattice.matrix
print("Vecteurs de la maille réciproque:", b1, b2, b3)
```

```python
from pymatgen.symmetry.analyzer import SpacegroupAnalyzer

#symétrie
analyzer = SpacegroupAnalyzer(structure)

# Identifier groupe ponctuel + système cristallin
spacegroup = analyzer.get_space_group_symbol()
crystal_system = analyzer.get_crystal_system()

print(f"Groupe ponctuel: {spacegroup}")
print(f"Système cristallin: {crystal_system}")
```
